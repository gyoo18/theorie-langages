```python
import re
```

```python
#La fonction accepte tous les charactères alphanumériques/tiret/underscore, peut contenir un point suivi d'autres charactèeres alphanumériques/tiret/underscore,
#suivi d'un @ obligatoire, contient un nom de domaine et un domaine premier
def validation_email(email):
  """
  Un courriel est structuré comme ceci :
    <nom>@<domaine>

  <nom> et <domaine> sont des suite de caractères alphanumériques avec des
  traits d'unions et des
  """
  validation_email = r'[a-zA-Z0-9]([\w_-]*[a-zA-Z0-9])?(\.[a-zA-Z0-9]([\w_-]*[a-zA-Z0-9])?)*@[a-zA-Z0-9]([\w_-]*[a-zA-Z0-9])?(\.[a-zA-Z0-9]([\w_-]*[a-zA-Z0-9])?)*'
  valide = re.match(validation_email, email)
  print("email valide" if valide else "email invalide")

#Le numéro accepté contient de 7 à 20 charactères numériques, espaces, parenthèses, tirets et points
def validation_telephone(telephone):
  numeros = re.sub(r'\D', '', telephone)
  if len(numeros) > 15:
    print(f"Numéro invalide")
  elif 15 >= len(numeros) > 10:
    print(f"+{numeros[-20:-10]} {numeros[-10:-7]} {numeros[-7:-4]} {numeros[-4:]}")
  else:
    print(f"{numeros[0:3]} {numeros[3:6]} {numeros[6:10]}")

def normaliser_code_postal(code):
    pattern = r'^\s*([ABCEGHJKLMNPRSTVWXYZ]\d[ABCEGHJKLMNPRSTVWXYZ])[\s-]*([0-9][ABCEGHJKLMNPRSTVWXYZ][0-9])\s*$'

    m = re.match(pattern, code.upper())
    if not m:
        return None

    return f"{m.group(1)} {m.group(2)}"


def main():
  validation_email("sfes.sefse@lsd.qc.ca")
  validation_telephone("(819) 492-1928")
  normaliser_code_postal(" h2y1c6 ")

main()
```
